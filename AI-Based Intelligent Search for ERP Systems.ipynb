```python
# ==========================================
# COMPLETE MISSING VALUE HANDLING
# ==========================================

import pandas as pd
import numpy as np

# ---------- ORDERS ----------

# Convert date columns
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_timestamp",
    "order_estimated_delivery_date"
]

for col in date_columns:
    orders[col] = pd.to_datetime(orders[col], errors="coerce")

# Handle missing approved date
orders["order_approved_at"] = orders["order_approved_at"].fillna(
    orders["order_purchase_timestamp"]
)

# DO NOT create fake delivery dates
# Create a meaningful delivery status
orders["delivery_status"] = np.where(
    orders["order_delivered_timestamp"].isna(),
    "Not Delivered",
    "Delivered"
)

# Flag for missing delivery timestamp
orders["delivery_date_missing"] = (
    orders["order_delivered_timestamp"].isna().astype(int)
)


# ---------- PRODUCTS ----------

# Missing category
products["product_category_name"] = products[
    "product_category_name"
].fillna("Unknown")

# Missing numerical product information
product_numeric_columns = [
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm"
]

for col in product_numeric_columns:
    products[col] = products[col].fillna(
        products[col].median()
    )


# ---------- CUSTOMERS ----------

# Handle possible missing text values
for col in ["customer_city", "customer_state"]:
    if col in customers.columns:
        customers[col] = customers[col].fillna("Unknown")


# ---------- ORDER ITEMS ----------

# Handle possible missing numerical values
for col in ["price", "shipping_charges"]:
    if col in order_items.columns:
        order_items[col] = order_items[col].fillna(
            order_items[col].median()
        )


# ---------- PAYMENTS ----------

# Handle possible missing categorical values
for col in ["payment_type"]:
    if col in payments.columns:
        payments[col] = payments[col].fillna("Unknown")

# Handle possible missing numerical values
for col in ["payment_sequential", "payment_installments", "payment_value"]:
    if col in payments.columns:
        payments[col] = payments[col].fillna(
            payments[col].median()
        )


# ==========================================
# FINAL MISSING VALUE CHECK
# ==========================================

print("========== MISSING VALUES AFTER CLEANING ==========")

print("\nCUSTOMERS:")
print(customers.isnull().sum())

print("\nORDER ITEMS:")
print(order_items.isnull().sum())

print("\nORDERS:")
print(orders.isnull().sum())

print("\nPAYMENTS:")
print(payments.isnull().sum())

print("\nPRODUCTS:")
print(products.isnull().sum())


# ==========================================
# DELIVERY STATUS SUMMARY
# ==========================================

print("\n========== DELIVERY STATUS ==========")
print(orders["delivery_status"].value_counts())

print("\n========== MISSING DELIVERY DATES ==========")
print(
    orders["order_delivered_timestamp"].isna().sum()
)

print("\nMissing value handling completed successfully! ✅")
```

```python
from sentence_transformers import SentenceTransformer

embedding_model = SentenceTransformer(
    "all-MiniLM-L6-v2"
)

print("Sentence Transformer loaded successfully!")
```

```python
query_embeddings = embedding_model.encode(
    query_df["clean_query"].tolist(),
    convert_to_numpy=True,
    show_progress_bar=True
)

print("Embedding shape:", query_embeddings.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

def find_similar_query(user_query, top_k=5):

    clean_query = preprocess_text(user_query)

    user_embedding = embedding_model.encode(
        [clean_query],
        convert_to_numpy=True
    )

    similarities = cosine_similarity(
        user_embedding,
        query_embeddings
    )[0]

    top_indices = similarities.argsort()[-top_k:][::-1]

    results = query_df.iloc[top_indices].copy()
    results["similarity"] = similarities[top_indices]

    return results
```

```python
dimension = query_embeddings.shape[1]

faiss_index = faiss.IndexFlatL2(dimension)

faiss_index.add(
    query_embeddings.astype("float32")
)

print("FAISS index created!")
print("Total vectors:", faiss_index.ntotal)
```

```python
def faiss_search(user_query, top_k=5):

    clean_query = preprocess_text(user_query)

    embedding = embedding_model.encode(
        [clean_query],
        convert_to_numpy=True
    ).astype("float32")

    distances, indices = faiss_index.search(
        embedding,
        top_k
    )

    results = query_df.iloc[indices[0]].copy()
    results["distance"] = distances[0]

    return results
```

```python
# ==========================================
# STEP 26 - SEMANTIC SIMILARITY
# ==========================================

from sklearn.metrics.pairwise import cosine_similarity

def semantic_search(user_query, top_k=5):
    
    clean_query = preprocess_text(user_query)
    
    user_embedding = embedding_model.encode(
        [clean_query],
        convert_to_numpy=True
    )
    
    similarity_scores = cosine_similarity(
        user_embedding,
        query_embeddings
    )[0]
    
    top_indices = similarity_scores.argsort()[-top_k:][::-1]
    
    results = query_df.iloc[top_indices].copy()
    
    results["similarity"] = similarity_scores[top_indices]
    
    return results
```

```python
# ==========================================
# STEP 37 - QUERY UNDERSTANDING
# ==========================================

import re

def extract_number(query):
    
    numbers = re.findall(
        r"\d+(?:\.\d+)?",
        query
    )
    
    if numbers:
        return float(numbers[0])
    
    return None
```

```python
print(extract_number("Show orders above 5000"))
print(extract_number("Find orders below 100"))
```

```python
# ==========================================
# STEP 38 - PAYMENT QUERY PARSER
# ==========================================

def detect_payment_type(query):
    
    query = query.lower()
    
    if "credit card" in query:
        return "credit_card"
    
    elif "debit card" in query:
        return "debit_card"
    
    elif "boleto" in query:
        return "boleto"
    
    elif "voucher" in query:
        return "voucher"
    
    elif "pix" in query:
        return "pix"
    
    return None
```

```python
print(
    detect_payment_type(
        "Show orders paid by credit card"
    )
)
```

```python
# ==========================================
# STEP 39 - AMOUNT FILTER
# ==========================================

def amount_filter(query):
    
    query = query.lower()
    
    amount = extract_number(query)
    
    if amount is None:
        return search_data
    
    # Above
    if "above" in query or "over" in query or "greater than" in query:
        
        result = search_data[
            search_data["total_payment"] > amount
        ]
    
    # Below
    elif "below" in query or "under" in query or "less than" in query:
        
        result = search_data[
            search_data["total_payment"] < amount
        ]
    
    else:
        result = search_data
    
    return result
```

```python
result = amount_filter(
    "Show orders above 5000"
)

print("Results:", len(result))

display(result.head(10))
```

```python
# ==========================================
# STEP 40 - PAYMENT FILTER
# ==========================================

def payment_filter(query):
    
    payment_type = detect_payment_type(query)
    
    if payment_type is None:
        return search_data
    
    result = search_data[
        search_data["payment_type"]
        .astype(str)
        .str.lower()
        == payment_type
    ]
    
    return result
```

```python
# ==========================================
# FIXED CUSTOMER CITY FILTER
# ==========================================

def customer_city_filter(query):
    
    query = preprocess_text(query)
    
    # Check whether customer_city exists
    if "customer_city" not in search_data.columns:
        print("customer_city column not found!")
        print("\nAvailable columns:")
        print(search_data.columns.tolist())
        return search_data.iloc[0:0]
    
    # Get unique cities
    cities = (
        search_data["customer_city"]
        .dropna()
        .astype(str)
        .str.lower()
        .str.strip()
        .unique()
    )
    
    # Find city mentioned in query
    for city in cities:
        
        if city in query:
            
            result = search_data[
                search_data["customer_city"]
                .astype(str)
                .str.lower()
                .str.strip()
                == city
            ]
            
            return result
    
    # If no city detected
    print("No matching city found in query.")
    return search_data.iloc[0:0]
```

```python
# ==========================================
# STEP 42 - PRODUCT CATEGORY FILTER
# ==========================================

def product_category_filter(query):
    
    query = query.lower()
    
    categories = search_data[
        search_data["product_category_name"]
        .astype(str)
        .str.lower()
        .unique()
    ]
    
    for category in categories:
        
        if category in query:
            
            result = search_data[
                search_data["product_category_name"]
                .astype(str)
                .str.lower()
                == category
            ]
            
            return result
    
    return search_data
```

```python
# ==========================================
# STEP 43 - ADVANCED SMART SEARCH
# ==========================================

def advanced_smart_search(user_query):
    
    query = preprocess_text(user_query)
    
    print("=" * 70)
    print("AI SMART ERP SEARCH")
    print("=" * 70)
    
    print("User Query:", user_query)
    
    # ----------------------------------
    # Delivery
    # ----------------------------------
    
    if "not delivered" in query or "undelivered" in query:
        
        result = search_data[
            search_data["delivery_status"] == "Not Delivered"
        ]
        
        intent = "delivery_search"
    
    
    elif "delivered" in query:
        
        result = search_data[
            search_data["delivery_status"] == "Delivered"
        ]
        
        intent = "delivery_search"
    
    
    # ----------------------------------
    # Cancelled
    # ----------------------------------
    
    elif "cancelled" in query or "canceled" in query:
        
        result = search_data[
            search_data["order_status"]
            .astype(str)
            .str.lower()
            == "canceled"
        ]
        
        intent = "status_search"
    
    
    # ----------------------------------
    # Amount
    # ----------------------------------
    
    elif (
        "above" in query
        or "below" in query
        or "over" in query
        or "under" in query
    ):
        
        result = amount_filter(query)
        
        intent = "amount_search"
    
    
    # ----------------------------------
    # Payment
    # ----------------------------------
    
    elif (
        "credit card" in query
        or "debit card" in query
        or "payment" in query
        or "paid" in query
    ):
        
        result = payment_filter(query)
        
        intent = "payment_search"
    
    
    # ----------------------------------
    # Customer City
    # ----------------------------------
    
    elif "customer" in query:
        
        result = customer_city_filter(query)
        
        intent = "customer_search"
    
    
    # ----------------------------------
    # Product Category
    # ----------------------------------
    
    elif "product" in query:
        
        result = product_category_filter(query)
        
        intent = "product_search"
    
    
    # ----------------------------------
    # General Order
    # ----------------------------------
    
    elif "order" in query:
        
        result = search_data
        
        intent = "order_search"
    
    
    else:
        
        result = search_data
        intent = "general_search"
    
    
    print("Detected Intent:", intent)
    print("Number of Results:", len(result))
    
    print("\nTop Results:")
    
    display(result.head(10))
    
    return result
```
